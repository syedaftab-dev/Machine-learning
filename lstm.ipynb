# Full Pipeline Summary

Raw Text


   ↓


Tokenization (word → index)
   ↓


Sequences
   ↓


Padding (same length)
   ↓


Embedding (dense vectors)
   ↓


LSTM (sequence learning)
   ↓


Dense (output)
   ↓


Prediction


```python
import numpy as np
import pandas as pd

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from sklearn.model_selection import train_test_split
```

```python
texts = [
    "I love this movie",
    "This film was terrible",
    "Amazing experience",
    "Worst acting ever",
    "I really liked it",
    "I hate this"
]

labels = [1, 0, 1, 0, 1, 0]  # 1 = positive, 0 = negative
```

```python
vocab_size = 1000
```

```python
tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer.fit_on_texts(texts)
```

```python
word_index = tokenizer.word_index
print(word_index)
```

```python
sequences = tokenizer.texts_to_sequences(texts)
print(sequences)
```

```python
# padding
max_length = 5
padded_sequences = pad_sequences(sequences, maxlen=max_length, padding="post", truncating="post")
print(padded_sequences)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    padded_sequences, labels, test_size=0.2, random_state=42
)
```

```python
embedding_dim = 16
model = Sequential([
    Embedding(input_dim=vocab_size,output_dim=embedding_dim, input_length=max_length),
    LSTM(32),
    Dense(1, activation="sigmoid")
])
```

```python
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train,
    np.array(y_train),
    epochs=10,
    validation_data=(X_test, np.array(y_test))
)
```

```python
loss, accuracy = model.evaluate(X_test, np.array(y_test))
print("Accuracy:", accuracy)
```

```python
test_sentence = ["I love this"]

seq = tokenizer.texts_to_sequences(test_sentence)
padded = pad_sequences(seq, maxlen=5, padding='post')

prediction = model.predict(padded)

print(prediction)  # closer to 1 → positive
```

🔥 Important Concepts (interview gold)

1️⃣ Tokenizer
Converts words → integers

2️⃣ Padding
Makes input fixed length

3️⃣ Embedding Layer
Learns word vectors automatically
Better than BoW / TF-IDF

4️⃣ LSTM
Captures sequence + context
