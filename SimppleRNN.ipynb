```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding, Dense, SimpleRNN, Input
from tensorflow.keras import Model
```

```python
sentences = [
    "I love this movie.",
    "The food was terrible.",
    "I am feeling very happy today.",
    "This phone is too expensive.",
    "The weather is nice.",
    "I hate waiting in long lines.",
    "She did a great job.",
    "The service was disappointing.",
    "I feel tired and bored.",
    "This book is amazing.",
    "The room is clean and quiet.",
    "I regret buying this product.",
    "He looks very excited.",
    "The exam was difficult.",
    "I enjoyed the trip a lot.",
    "The coffee tastes bad.",
    "Everything is going well.",
    "I am upset with the result.",
    "The dress looks beautiful.",
    "This app keeps crashing.",
    "I am proud of my team.",
    "The train was late again.",
    "She is very kind and helpful.",
    "I do not like this place.",
    "The music is relaxing.",
    "I feel nervous about tomorrow.",
    "The package arrived on time.",
    "This laptop is slow and noisy.",
    "I am satisfied with the service.",
    "Nothing special happened today."
]
labels = [0]*15 + [1]*15
labels = np.array(labels)
vocab_size = 2000
```

```python
tok = Tokenizer(num_words=vocab_size, oov_token='<OOV>')
tok.fit_on_texts(sentences)
seqs = tok.texts_to_sequences(sentences)
max_len = max([len(seq) for seq in seqs])
X = pad_sequences(seqs, maxlen=max_len, padding='post')
```

```python
y = labels
```

```python
embed_dim=16
rnn_units=8
```

```python
inp = Input(shape=(max_len,),dtype='int32',name='input')
```

```python
x = Embedding(input_dim=vocab_size, output_dim=embed_dim,mask_zero=True,name='embed')(inp)
```

```python
rnn = SimpleRNN(units=rnn_units,return_sequences=False,name='simple_rnn')
```

```python
x_last = rnn(x)
```

```python
out = Dense(1,activation='sigmoid',name='out')(x_last)
```

```python
model = Model(inputs=inp,outputs=out)
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,y,epochs=25,batch_size=25,verbose=1)
```
